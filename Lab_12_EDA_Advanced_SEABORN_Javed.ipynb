```python
df = pd.read_excel("Apex_Retail_Business_US (1).xlsx")
```

```python
df.head()
```

Question-1: How are sales distributed

```python
sns.histplot(data=df,x="Sales",kde=True)
plt.title("Sales Distribution")
plt.show()
```

Question-2: Which product category showed the greatest variation in sales?

```python
sns.boxplot(data=df, x="Category",y="Sales")
plt.title("Sales Distribution by Category")
plt.show()
```

Question-3:which region has the higest profit?

```python
sns.barplot(data=df,x="Region",y="Profit")
plt.title("Average Profit by Region")
plt.show()
```

Q4: what is the distribution of the customer rating?

```python
sns.countplot(data=df,x="Customer_Rating")
plt.title("Customer Rating distribution")
plt.show()
```

Question-5: is there is reltionship between Sales and Profit?

```python
sns.regplot(data=df,x="Sales", y="Profit")
plt.title("Sales vs profit")
plt.show()
```

Q6:Which Payment mode is used more frequently?

```python
sns.countplot(data=df,x="Payment_Mode")
plt.title("Payment Mode Usage")
plt.show()
```

Q7:Which Variable are stronglycorelated?


```python
correlation=df[["Sales","Profit","Quantity","Unit_Price","Discount","Customer_Rating"]].corr()
sns.heatmap(correlation,annot=True,cmap="RdPu")
plt.title("CorrelationHeatmap")
plt.show()
```

Q8: Does Discount affect profit?

```python
sns.scatterplot(data=df,x="Discount",y="Profit")
plt.title("Discount vs Profit")
plt.show()
```
